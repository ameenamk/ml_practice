```python
from sklearn.model_selection import train_test_split
X=df.drop("HeartAttack",axis=1)
y=df["HeartAttack"].map({"Yes":1,"No":0})
X_train,X_test,y_train,y_test=train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=0,
    stratify=y
)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder,StandardScaler,MinMaxScaler

preprocessor=ColumnTransformer(
    transformers=[
        ("Gender",Pipeline([
            ("imputer",SimpleImputer(strategy="most_frequent")),
            ("encoder",OrdinalEncoder())
        ]),["Gender"]),
        ("Age",Pipeline([
            ("imputer",SimpleImputer(strategy="mean")),
            ("scaler",StandardScaler())
        ]),["Age"]),
        ("HasTension",OrdinalEncoder(),["HasTension"]),
        ("AnyHeartDisease",OrdinalEncoder(),["AnyHeartDisease"]),
        ("NeverMarried",OrdinalEncoder(),["NeverMarried"]),
        ("Occupation",OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ),["Occupation"]),
        
        # 7. LivesIn
        ("LivesIn", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OrdinalEncoder())
        ]), ["LivesIn"]),

        # 8. GlucoseLevel
        ("GlucoseLevel", Pipeline([
            ("imputer", SimpleImputer(strategy="mean")),
            ("scaler", MinMaxScaler())
        ]), ["GlucoseLevel"]),

        # 9. BMI
        ("BMI", Pipeline([
            ("imputer", SimpleImputer(strategy="mean")),
            ("scaler", StandardScaler())
        ]), ["BMI"]),

        # 10. SmokingStatus
        ("SmokingStatus", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ))
        ]), ["SmokingStatus"])
    ],
    remainder="drop"
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
```

## Model Building

```python
from sklearn.model_selection import train_test_split

X = md.iloc[:, :-1]
y = md.iloc[:, -1]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    shuffle=False
)
```

```python
from sklearn.linear_model import Perceptron
model=Perceptron(
    random_state=1729,
    eta0=1.0,
    max_iter=5,
    shuffle=False,
    fit_intercept=True,
    validation_fraction=0.1,
    penalty=None,
    warm_start=True   
)
model.fit(X_train,y_train)
print(model.intercept_)
```

```python
from sklearn.metrics import precision_score

# 1. Generate predictions on the training data using your trained model
y_pred_train = model.predict(X_train)

# 2. Compute precision for the positive class (pos_label=1)
precision = precision_score(y_train,y_pred_train, pos_label=1)

print(precision)
```

```python
import numpy as np
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import log_loss

model = SGDClassifier(
    loss='log_loss',
    penalty='l2',
    eta0=0.001,
    alpha=0,
    learning_rate='constant',
    random_state=1729,
    warm_start=True,
    max_iter=1,
    shuffle=False
)

losses = []

for i in range(5):
    model.partial_fit(X_train, y_train, classes=np.unique(y_train))
    
    y_prob = model.predict_proba(X_train)
    loss = log_loss(y_train, y_prob)
    losses.append(loss)
    
    print(f"Iteration {i+1}: {loss:.3f}")
```

```python
from sklearn.linear_model import SGDClassifier
from sklearn.model_selection import GridSearchCV

# SGD Classifier
sgd = SGDClassifier(
    loss='log_loss',
    learning_rate='constant',
    random_state=1729
)

# Parameters to search
param_grid = {
    'alpha': [0.0001, 0.0005, 0.001, 0.005],
    'eta0': [0.01, 0.05, 0.1, 0.5]
}

# Grid Search
grid = GridSearchCV(
    estimator=sgd,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy'
)

# Train using TRAINING DATA ONLY
grid.fit(X_train, y_train)

print("Best parameters:", grid.best_params_)
print("Best score:", grid.best_score_)
```

```python
from sklearn.linear_model import SGDClassifier

# Get best parameters from previous GridSearch
best_alpha = grid.best_params_['alpha']
best_eta0 = grid.best_params_['eta0']

# Create new estimator
model_weighted = SGDClassifier(
    loss='log_loss',
    learning_rate='constant',
    random_state=1729,
    alpha=best_alpha,
    eta0=best_eta0,
    class_weight={0: 0.1, 1: 2}
)

# Train ONLY on training data
model_weighted.fit(X_train, y_train)

# Predict test data
y_test_pred = model_weighted.predict(X_test)
```

```python
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Base estimators
dt = DecisionTreeClassifier(random_state=1729)

knn = KNeighborsClassifier()

lr = LogisticRegression(random_state=1729)

# Bagging models
bag_dt = BaggingClassifier(
    estimator=dt,
    n_estimators=20,
    random_state=1729
)

bag_knn = BaggingClassifier(
    estimator=knn,
    n_estimators=20,
    random_state=1729
)

bag_lr = BaggingClassifier(
    estimator=lr,
    n_estimators=20,
    random_state=1729
)

# Train
bag_dt.fit(X_train, y_train)
bag_knn.fit(X_train, y_train)
bag_lr.fit(X_train, y_train)

# Predictions
dt_train_pred = bag_dt.predict(X_train)
dt_test_pred = bag_dt.predict(X_test)

knn_train_pred = bag_knn.predict(X_train)
knn_test_pred = bag_knn.predict(X_test)

lr_train_pred = bag_lr.predict(X_train)
lr_test_pred = bag_lr.predict(X_test)

# Accuracy
dt_train = accuracy_score(y_train, dt_train_pred)
dt_test = accuracy_score(y_test, dt_test_pred)

knn_train = accuracy_score(y_train, knn_train_pred)
knn_test = accuracy_score(y_test, knn_test_pred)

lr_train = accuracy_score(y_train, lr_train_pred)
lr_test = accuracy_score(y_test, lr_test_pred)

# Absolute train-test difference
dt_diff = abs(dt_train - dt_test)
knn_diff = abs(knn_train - knn_test)
lr_diff = abs(lr_train - lr_test)

print("Decision Tree:")
print("Train:", dt_train, "Test:", dt_test, "Difference:", dt_diff)

print("\nKNN:")
print("Train:", knn_train, "Test:", knn_test, "Difference:", knn_diff)

print("\nLogistic Regression:")
print("Train:", lr_train, "Test:", lr_test, "Difference:", lr_diff)
```

```python
differences = {
    "Decision Tree": dt_diff,
    "KNN": knn_diff,
    "Logistic Regression": lr_diff
}

print(min(differences, key=differences.get))
```

```python
from sklearn.ensemble import VotingClassifier
from sklearn.metrics import accuracy_score

# Individual classifiers with the same settings
dt = DecisionTreeClassifier(random_state=1729)

knn = KNeighborsClassifier()

lr = LogisticRegression(random_state=1729)

# Voting Classifier
voting = VotingClassifier(
    estimators=[
        ('dt', dt),
        ('knn', knn),
        ('lr', lr)
    ],
    voting='hard'
)

# Train on training data
voting.fit(X_train, y_train)

# Predictions
y_train_pred = voting.predict(X_train)
y_test_pred = voting.predict(X_test)

# Accuracy scores
train_score = accuracy_score(y_train, y_train_pred)
test_score = accuracy_score(y_test, y_test_pred)

# Absolute difference
difference = abs(train_score - test_score)

print("Train score:", train_score)
print("Test score:", test_score)
print("Absolute difference:", round(difference, 4))
```
