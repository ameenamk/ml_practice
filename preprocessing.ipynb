### Create a column 'CATEGORY' and divide the houses in categories a following
* 1. house price < 10 lacs
  2. 10 <= house price < 20lacs
  3. 20 <= house price < 30 lacs
  4. 30 <= house price < 40lacs
  5. house price >- 40lacs

### which category has the highest no of records

```python
bins=[-float('inf'),10,20,30,40,float('inf')]
labels=['c1','c2','c3','c4','c5']
data['CATEGORY']=pd.cut(data['PRICE'],bins=bins,labels=labels,right=False)
data['CATEGORY'].value_counts().idxmax()
```
